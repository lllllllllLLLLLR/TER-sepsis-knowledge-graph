# Sepsis Papers: PDF Content Extraction

This notebook automates the process of extracting raw text content from research papers stored in PDF format. It scans the subdirectories for diagnosis and treatment papers and compiles the data into a structured Excel file.

**Workflow:**
1. **Directory Scanning**: Iterates through `Sepsis_Papers/Diagnosis` and `Sepsis_Papers/Treatment`.
2. **Text Extraction**: Uses the `PyMuPDF` library to perform a raw text dump from every page of the PDFs.
3. **Data Structuring**: Captures the original paper code (filename) alongside the full text.
4. **Export**: Saves the consolidated dataset to a `.xlsx` file for use in the LLM analysis pipeline.

```python
import fitz
import pandas as pd
import re
from pathlib import Path

SOURCE_DIR = Path("Sepsis_Papers")
SUB_DIRS = ["Diagnosis", "Treatment"]
OUTPUT_FILE = "Text_Papers.xlsx"
 
def clean_for_excel(text):
    if not text:
        return ""
    illegal_chars = re.compile(r'[\000-\010]|[\013-\014]|[\016-\037]')
    return illegal_chars.sub("", text)

all_data = []

for folder_name in SUB_DIRS:
    folder_path = SOURCE_DIR / folder_name
    
    for pdf_file in folder_path.glob("*.pdf"):
        paper_code = pdf_file.stem
        raw_text = ""
        
        try:
            with fitz.open(pdf_file) as doc:
                for page in doc:
                    raw_text += page.get_text()
            
            # Clean the text before adding it to the list
            cleaned_text = clean_for_excel(raw_text)
            
            all_data.append({
                "Paper_Code": paper_code,
                "Category": folder_name,
                "Content": cleaned_text.strip()
            })
            
        except Exception as e:
            print(f"Failed to read {pdf_file.name}: {e}")

df = pd.DataFrame(all_data)

# Save to Excel
try:
    df.to_excel(OUTPUT_FILE, index=False, engine='openpyxl')
    print(f"Processing Complete. File saved to: {OUTPUT_FILE}")
except Exception as e:
    print(f"Excel save failed: {e}")
```
